1. PyPREP
Digunakan untuk membersihkan dan mempersiapkan data sinyal EEG sebelum dianalisis. Misalnya mengurangi noise dan memperbaiki kualitas sinyal EEG.
2. SciPy
Digunakan untuk perhitungan ilmiah dan matematika dalam Python. Fungsinya mencakup statistik, aljabar linear, optimasi, serta pemrosesan sinyal.
3. Magic Wand (Tongkat Sihir)
Digunakan untuk mengenali gerakan atau gesture berdasarkan data sensor seperti accelerometer dan gyroscope. Contohnya membedakan gerakan mengayunkan atau menggerakkan suatu perangkat.
4. pyECG
Digunakan untuk membaca, mengolah, dan bekerja dengan data ECG (Electrocardiogram), yaitu data yang merekam aktivitas listrik jantung. Library ini dapat membantu dalam pengolahan data ECG dari berbagai format dataset.
